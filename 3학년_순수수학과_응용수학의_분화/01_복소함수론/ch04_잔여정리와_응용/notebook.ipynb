## 이론 정리

### 정의와 관점
- 복소함수는 평면에서 평면으로 가는 함수이며, 해석성은 복소미분 가능성이 열린 영역에서 성립하는 강한 조건이다.
- 복소미분은 모든 방향에서 같은 선형 근사가 존재해야 하므로 실변수 미분보다 훨씬 강하다.
- 특이점은 함수가 해석적이지 않은 점이며, 극점과 제거가능 특이점 등으로 분류된다.
- 등각사상은 국소적으로 각을 보존하는 복소함수의 기하적 성질이다.

### 핵심 명제와 정리
- Cauchy 적분정리는 해석함수의 폐곡선 적분이 0이 되는 조건을 제공한다.
- Cauchy 적분공식은 내부 함수값이 경계값에 의해 결정됨을 보여준다.
- Laurent 전개와 잔여정리는 특이점 주변의 한 계수로 복소적분을 계산하게 한다.
- 해석함수는 실수부와 허수부가 조화함수이며 물리적 포텐셜 문제와 연결된다.

### 계산과 학습 절차
- 영역, 경계, 특이점의 위치를 먼저 그림으로 표시한다.
- 함수가 해석적인 영역을 찾고, 적분 경로가 특이점을 감싸는지 확인한다.
- 급수 전개를 쓸 때는 중심과 수렴환을 명확히 둔다.
- 복소사상은 격자 변환을 그려 각 보존과 왜곡을 함께 확인한다.

### 자주 생기는 오해
- 실수부와 허수부가 각각 미분 가능해도 Cauchy-Riemann 조건과 연속성이 필요하다.
- 적분값은 경로와 특이점 위치에 민감하므로 단순히 원시함수만 찾으면 안 되는 경우가 있다.
- 잔여정리는 특이점을 빠짐없이 분류해야 정확히 적용된다.

### 증명으로 연결하기
- 복소해석 증명은 삼각형 또는 원판에서의 국소 적분 성질을 전체 영역으로 확장한다.
- 급수 전개는 균등수렴을 이용해 미분과 적분을 항별로 정당화한다.
- 많은 정리가 해석성 하나에서 매우 강한 결론을 뽑아내므로 가정의 영역을 항상 확인한다.

## 0. 실행 준비

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("3학년_순수수학과_응용수학의_분화/01_복소함수론/ch04_잔여정리와_응용")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "contour_integral"
TITLE = "복소함수론 - 잔여정리와 응용"
CONCEPT_EXAMPLE = "극점 주변에서 적분값이 국소 정보로 결정됨을 시각화한다."
APPLICATION_EXAMPLE = "진동 적분을 빠르게 평가하는 복소해석 도구를 이해한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
