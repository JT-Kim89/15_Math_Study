## 이론 정리

### 정의와 관점
- 최적화는 목적함수를 제약조건 아래에서 최소화하거나 최대화하는 문제다.
- 볼록집합과 볼록함수는 국소 최적해가 전역 최적해가 되는 강력한 구조를 제공한다.
- 그래디언트는 목적함수의 가장 가파른 증가 방향이고, 음의 그래디언트는 하강 방향이다.
- 선형계획법은 선형 목적함수와 선형 제약으로 이루어진 최적화 문제다.

### 핵심 명제와 정리
- 볼록 최적화에서는 1차 조건이 전역 최적성을 보장할 수 있다.
- KKT 조건은 제약최적화에서 최적점이 만족해야 하는 일반화된 미분 조건이다.
- 쌍대성은 원문제의 최적값을 다른 문제의 하한 또는 상한으로 해석하게 해준다.
- 선형계획법의 최적해는 가능영역의 꼭짓점에서 찾을 수 있다.

### 계산과 학습 절차
- 목적함수, 변수, 제약, 가능한 영역을 명확히 적는다.
- 볼록성, 미분가능성, 제약자격조건을 확인해 사용할 정리를 고른다.
- 반복 알고리즘은 학습률, 정지조건, 목적값 감소 여부를 기록한다.
- 선형계획 문제는 그림 가능한 2차원 예제로 꼭짓점과 쌍대해석을 먼저 익힌다.

### 자주 생기는 오해
- 비볼록 문제에서는 그래디언트가 0인 점이 전역 최적해가 아닐 수 있다.
- 학습률이 너무 크면 하강법이 발산하거나 진동한다.
- 쌍대 간격이 0이 되려면 조건이 필요하며 항상 강한 쌍대성이 성립하지는 않는다.

### 증명으로 연결하기
- 볼록성 증명은 Jensen 부등식, 접평면 부등식, Hessian 양의 준정부호성을 이용한다.
- KKT 조건은 Lagrangian과 법선원뿔 관점에서 도출된다.
- 알고리즘 수렴 증명은 목적값 감소, Lipschitz 조건, 강볼록성 같은 가정을 사용한다.

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("3학년_순수수학과_응용수학의_분화/11_선형계획법/ch01_가능영역")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "linear_programming"
TITLE = "선형계획법 - 가능영역"
CONCEPT_EXAMPLE = "2변수 선형부등식이 만드는 다각형을 그린다."
APPLICATION_EXAMPLE = "노동시간과 원료 제한 아래 최대 이익 조합을 찾는다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
