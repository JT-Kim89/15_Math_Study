## 이론 정리

### 정의와 관점
- 측도론은 길이, 면적, 확률을 하나의 추상적 크기 함수로 통합한다.
- 시그마 대수는 측도를 부여할 수 있는 사건 또는 집합들의 안정적인 모음이다.
- 가측함수는 함수값의 역상이 가측집합이 되는 함수다.
- Lebesgue 적분은 함수의 정의역 분할보다 함수값 수준집합의 크기를 이용해 적분을 정의한다.

### 핵심 명제와 정리
- 단조수렴정리와 지배수렴정리는 극한과 적분을 교환할 수 있는 강력한 조건을 제공한다.
- Fatou 보조정리는 극한하한과 적분 사이의 기본 부등식을 준다.
- 거의 어디서나 성질은 측도 0인 예외를 무시하고 함수공간을 다루게 해준다.
- Lebesgue 적분은 Riemann 적분보다 더 넓은 함수와 수렴 현상을 안정적으로 다룬다.

### 계산과 학습 절차
- 먼저 어떤 집합에 측도를 줄 수 있는지, 어떤 함수가 가측인지 확인한다.
- 적분 문제는 단순함수 근사, 양/음 부분 분해, 수렴정리 적용 가능성 순서로 본다.
- 확률 문제에서는 사건의 시그마 대수와 확률측도를 명시한다.
- 수렴은 점별, 거의 어디서나, Lp, 측도수렴을 구분해 기록한다.

### 자주 생기는 오해
- 모든 부분집합이 항상 가측인 것은 아니며, 시그마 대수 선택이 중요하다.
- 점별수렴만으로 적분 극한 교환은 보장되지 않는다.
- 측도 0 집합에서의 함수값 차이는 Lp 공간에서는 같은 원소로 취급될 수 있다.

### 증명으로 연결하기
- 측도론 증명은 단순함수에서 시작해 단조극한으로 일반 함수에 확장한다.
- 수렴정리는 양의 함수, 절댓값 지배, 균등적분가능성 같은 조건을 통해 정당화된다.
- 확률론의 엄밀한 증명은 측도론의 언어 위에서 조건부기댓값과 적분을 사용한다.
