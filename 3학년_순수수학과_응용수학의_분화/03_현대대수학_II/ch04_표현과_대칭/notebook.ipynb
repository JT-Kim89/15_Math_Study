## 이론 정리

### 정의와 관점
- 대수학은 수, 대칭, 연산을 공리적 구조로 추상화한다.
- 군은 하나의 연산과 역원을 가진 대칭 구조이고, 환은 덧셈과 곱셈을 함께 가진 구조다.
- 체는 0이 아닌 원소가 모두 곱셈 역원을 가져 나눗셈이 가능한 구조다.
- 동형은 이름은 달라도 연산 구조가 같은 대상을 식별한다.

### 핵심 명제와 정리
- Lagrange 정리는 유한군에서 부분군의 크기가 전체 군의 크기를 나눈다는 강한 제약을 준다.
- 군 작용은 대칭이 집합 위에서 어떻게 움직이는지를 궤도와 안정자로 분해한다.
- 몫구조는 같은 것으로 보고 싶은 원소들을 묶어 더 단순한 대수 구조를 만든다.
- 유한체는 암호, 오류정정, 조합 설계에서 계산 가능한 대수 모델을 제공한다.

### 계산과 학습 절차
- 연산이 닫혀 있는지, 항등원과 역원이 있는지, 결합법칙이 성립하는지 확인한다.
- 구조를 이해할 때는 작은 Cayley 표, 생성원, 부분구조를 먼저 계산한다.
- 다항식 문제는 계수 체가 달라질 때 근과 분해가 어떻게 바뀌는지 비교한다.
- 동형을 보일 때는 연산 보존과 전단사성을 동시에 확인한다.

### 자주 생기는 오해
- 연산표가 비슷해 보여도 항등원이나 역원이 없으면 군이 아니다.
- 정수에서 성립하는 직관이 mod p나 다항식 환에서는 달라질 수 있다.
- 몫구조에서는 대표원 선택이 계산에는 필요하지만 결과는 대표원에 의존하면 안 된다.

### 증명으로 연결하기
- 대수 증명은 공리를 직접 사용해 구조가 보존됨을 확인하는 방식이 많다.
- 동형정리는 핵심 정보가 핵과 상에 어떻게 압축되는지 보여준다.
- 작은 예제 계산은 정리의 가정이 왜 필요한지 확인하는 가장 좋은 출발점이다.
