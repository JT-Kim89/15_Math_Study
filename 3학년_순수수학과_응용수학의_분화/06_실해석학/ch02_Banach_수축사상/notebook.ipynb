## 이론 정리

### 정의와 관점
- 실해석은 미적분의 직관을 실수의 완비성과 극한의 엄밀한 언어로 다시 세우는 과목이다.
- 수열의 수렴, Cauchy 조건, 상한/하한은 실수계 구조의 핵심 표현이다.
- 거리공간은 가까움과 수렴을 추상화해 함수공간이나 기하 공간에도 같은 논리를 적용한다.
- 균등수렴은 모든 점에서 동시에 오차를 통제하는 강한 수렴 개념이다.

### 핵심 명제와 정리
- 단조수렴정리와 Bolzano-Weierstrass 정리는 유계성과 완비성이 수렴 부분열을 만든다는 사실을 말한다.
- Heine-Borel 정리는 유클리드 공간에서 닫힘과 유계성이 compactness와 같아지는 특별한 결과다.
- 균등수렴은 연속성, 적분, 때로는 미분과 극한의 교환을 정당화한다.
- Banach 고정점 정리는 수축사상이 유일한 고정점과 반복수렴을 보장한다.

### 계산과 학습 절차
- 정의의 양화사 순서를 먼저 분석한다. 특히 epsilon, N, delta, x의 순서가 중요하다.
- 증명 전에는 반례 후보를 찾아 어떤 가정이 필요한지 감각을 만든다.
- 함수열은 점별 오차와 전체 구간 최대 오차를 따로 계산한다.
- 거리공간 문제는 열린공, 닫힌집합, 수렴열 중 어떤 관점이 가장 짧은지 고른다.

### 자주 생기는 오해
- 점별수렴은 균등수렴보다 약하며, 연속성이나 적분 교환을 자동 보장하지 않는다.
- 닫힌집합과 compact 집합은 일반 거리공간에서 같은 개념이 아니다.
- 직관적으로 당연한 극한 교환은 대개 추가 조건 없이는 거짓이다.

### 증명으로 연결하기
- 해석학 증명은 정의를 정확히 펼치고, 원하는 epsilon에 맞게 N이나 delta를 구성하는 방식이 많다.
- compactness 증명은 열린덮개, 수열 compactness, 유한 부분덮개 사이의 변환을 자주 사용한다.
- 반례는 보통 경계점, 무한구간, 진동, 뾰족한 함수열에서 나온다.
