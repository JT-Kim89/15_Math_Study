## 실제 응용 예제
조사 표본 수가 오차에 미치는 영향을 설명한다.

## 이론 정리

### 정의와 관점
- 확률공간은 표본공간, 사건, 확률측도로 이루어진 불확실성의 수학적 모델이다.
- 확률변수는 무작위 결과를 수치로 바꾸는 함수이며, 분포는 그 함수값의 확률 구조다.
- 조건부확률은 정보가 주어졌을 때 표본공간을 다시 보는 방식이다.
- 기댓값은 장기 평균 또는 분포에 대한 적분으로 해석된다.

### 핵심 명제와 정리
- Bayes 공식은 관측된 증거를 이용해 가설의 확률을 갱신한다.
- 큰수의 법칙은 표본평균이 참 평균으로 안정됨을 설명한다.
- 중심극한정리는 많은 독립 효과의 평균이 정규분포로 근사되는 이유를 준다.
- Markov 성질은 미래가 현재 상태에만 의존하는 동역학을 단순화한다.

### 계산과 학습 절차
- 표본공간과 사건을 먼저 정의하고, 독립성 가정이 실제로 타당한지 확인한다.
- 시뮬레이션은 닫힌형 공식이 어려울 때 분포와 기대값을 탐색하는 도구로 쓴다.
- 조건부 문제는 분모가 되는 조건 사건의 확률이 0이 아닌지 확인한다.
- 반복 실험에서는 표본 수를 늘리며 추정량의 변동성이 어떻게 줄어드는지 본다.

### 자주 생기는 오해
- 독립과 배반은 다른 개념이며, 조건부확률은 직관과 반대로 움직일 수 있다.
- 표본이 작으면 경험적 빈도는 참 확률과 크게 다를 수 있다.
- 정규근사는 편리하지만 꼬리확률이나 작은 표본에서는 부정확할 수 있다.

### 증명으로 연결하기
- 확률 증명은 지시함수, 조건부기댓값, 분해공식으로 복잡한 사건을 쪼갠다.
- 극한정리는 독립성, 동일분포, 분산 유한성 같은 조건을 꼼꼼히 확인해야 한다.
- 시뮬레이션은 정리를 증명하지 않지만 가정이 깨졌을 때 어떤 현상이 생기는지 보여준다.

## 0. 실행 준비

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("3학년_순수수학과_응용수학의_분화/09_수리통계학/ch02_표본분포")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "clt"
TITLE = "수리통계학 - 표본분포"
CONCEPT_EXAMPLE = "같은 모집단에서 많은 표본평균을 뽑아 분포를 만든다."
APPLICATION_EXAMPLE = "조사 표본 수가 오차에 미치는 영향을 설명한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
