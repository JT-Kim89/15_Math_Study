## 실제 응용 예제
고객 만족도 평균을 작은 표본으로 추정한다.

## 이론 정리

### 정의와 관점
- 통계학은 데이터가 확률모형에서 왔다고 가정하고 미지의 파라미터나 미래 관측을 추론한다.
- 추정량은 표본에서 파라미터를 계산하는 규칙이며, 편향과 분산으로 성질을 평가한다.
- 가능도는 관측된 데이터가 주어졌을 때 파라미터가 얼마나 그럴듯한지를 나타낸다.
- 베이지안 통계는 사전분포와 가능도를 결합해 사후분포를 만든다.

### 핵심 명제와 정리
- 최대가능도추정은 큰 표본에서 일관성과 점근정규성을 갖는 경우가 많다.
- 신뢰구간은 반복표본에서 일정 비율로 참값을 포함하도록 설계된 절차다.
- 가설검정은 귀무가설 아래 관측값이 얼마나 극단적인지 p-value로 측정한다.
- 공액사전분포는 사후분포 계산을 같은 분포족 안에서 닫히게 해준다.

### 계산과 학습 절차
- 데이터 생성 과정, 독립성, 표본추출 방식, 관측오차를 먼저 명시한다.
- 모형을 세운 뒤 추정량, 표준오차, 잔차, 진단 그래프를 함께 본다.
- 검정에서는 효과크기와 불확실성을 p-value와 함께 해석한다.
- 베이지안 분석은 사전분포 민감도와 사후예측검사를 확인한다.

### 자주 생기는 오해
- 상관은 인과를 의미하지 않으며, 누락변수와 선택편향이 해석을 왜곡할 수 있다.
- p-value는 귀무가설이 참일 확률이 아니다.
- 작은 표본에서는 정규근사와 점근이론을 조심해야 한다.

### 증명으로 연결하기
- 통계 이론은 추정량의 분포, 기대값, 분산, 극한분포를 계산하는 데서 출발한다.
- 가능도 기반 증명은 로그가능도의 미분, 정보행렬, Taylor 전개를 자주 사용한다.
- 베이지안 계산은 Bayes 공식과 적분 정규화 상수를 중심으로 전개된다.

## 0. 실행 준비

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("1학년_기초_수학과_증명_입문/08_기초통계학/ch03_추정과_신뢰구간")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "distribution_inference"
TITLE = "기초통계학 - 추정과 신뢰구간"
CONCEPT_EXAMPLE = "표본평균의 신뢰구간이 표본크기에 따라 좁아짐을 확인한다."
APPLICATION_EXAMPLE = "고객 만족도 평균을 작은 표본으로 추정한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
