## 실제 응용 예제
지형의 고도 자료에서 가장 가파른 상승 방향을 찾는다.

## 이론 정리

### 정의와 관점
- 여러 변수 함수는 한 방향의 변화가 아니라 모든 방향에서의 변화율 구조를 함께 가진다.
- 편미분은 좌표축 방향 변화율이고, 방향도함수는 임의 방향 변화율이다.
- 그래디언트는 가장 빠르게 증가하는 방향을 가리키며 등고선과 직교한다.
- 중적분은 면적 또는 부피 조각의 합을 극한으로 보낸 누적량이다.

### 핵심 명제와 정리
- 연쇄법칙은 여러 변수의 의존 관계를 미분 가능한 합성 구조로 정리한다.
- Jacobian은 좌표변환에서 국소적인 면적/부피 배율을 나타낸다.
- Green, Stokes, 발산정리는 경계에서의 적분과 내부의 미분량을 연결한다.
- Hessian은 임계점의 국소적 형태와 최적화 성질을 판별하는 핵심 도구다.

### 계산과 학습 절차
- 등고선, 그래디언트, 임계점을 먼저 그려 함수의 지형을 파악한다.
- 좌표변환을 쓸 때는 영역, 경계, Jacobian을 한 묶음으로 확인한다.
- 벡터장은 화살표 그림만 보지 말고 발산과 회전을 함께 계산한다.
- 수치 격자를 만들 때는 x, y 방향 간격과 경계 처리를 명시한다.

### 자주 생기는 오해
- 편미분이 존재한다고 해서 항상 연속이나 전미분가능성이 보장되지는 않는다.
- 좌표변환에서 Jacobian 절댓값을 빠뜨리면 누적량 해석이 틀어진다.
- 벡터장 그림은 스케일링 방식에 따라 직관을 과장할 수 있다.

### 증명으로 연결하기
- 다변수 증명은 보통 선형근사와 잔차항의 크기 추정으로 전개한다.
- 적분정리는 작은 셀에서의 국소 관계를 전체 영역에 더한 뒤 내부 경계가 상쇄되는 구조를 사용한다.
- 최적화 정리는 1차 조건과 2차 조건을 분리해 필요조건과 충분조건을 구분한다.
