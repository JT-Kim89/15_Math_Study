## 이론 정리

### 정의와 관점
- 위상수학은 거리나 각도보다 연속성, 연결성, 구멍, 변형 가능성을 연구한다.
- 위상공간은 열린집합을 기본 데이터로 삼아 수렴과 연속을 정의한다.
- 몫공간은 일부 점들을 같은 점으로 붙여 새로운 공간을 만든다.
- 단체복합체와 호몰로지는 공간의 구멍을 계산 가능한 대수 자료로 바꾼다.

### 핵심 명제와 정리
- 연속함수는 연결성, compactness 같은 위상적 성질을 보존한다.
- 기본군은 루프가 서로 연속적으로 변형될 수 있는지를 측정한다.
- 호몰로지는 차원별 구멍의 수와 구조를 선형대수 계산으로 추출한다.
- Euler 지표는 꼭짓점, 변, 면의 조합으로 공간의 위상 정보를 압축한다.

### 계산과 학습 절차
- 공간을 점, 열린집합, 경계 붙이기, 루프 관점 중 하나로 표현한다.
- 연속변형으로 보존되는 성질과 보존되지 않는 성질을 분리한다.
- 데이터나 메시에서는 단체복합체를 만들고 차원별 셀 수와 연결성을 계산한다.
- 그림을 그릴 때는 늘어남과 휘어짐은 허용되지만 찢기와 붙이기는 다른 연산임을 기억한다.

### 자주 생기는 오해
- 모양이 비슷해 보여도 구멍 수나 경계 붙임 방식이 다르면 위상적으로 다를 수 있다.
- 거리공간의 직관을 일반 위상공간에 그대로 적용하면 틀릴 수 있다.
- 몫공간은 붙인 뒤의 열린집합 정의가 중요하며 그림만으로 판단하면 위험하다.

### 증명으로 연결하기
- 위상 증명은 열린집합의 원상, 덮개, 연속변형을 정의 그대로 추적한다.
- 불변량은 두 공간이 같지 않음을 보일 때 강력하지만, 같음을 보이려면 실제 사상을 구성해야 한다.
- 대수적 위상수학은 공간 문제를 군, 가군, 사슬복합체 계산으로 변환한다.

## 0. 실행 준비

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("4학년_심화_과목과_연구_주제/07_대수적_위상수학/ch02_기본군")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "topology_space"
TITLE = "대수적 위상수학 - 기본군"
CONCEPT_EXAMPLE = "고리 모양 공간에서 수축되지 않는 루프를 그린다."
APPLICATION_EXAMPLE = "장애물이 있는 로봇 경로 공간의 서로 다른 루프 유형을 구분한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
