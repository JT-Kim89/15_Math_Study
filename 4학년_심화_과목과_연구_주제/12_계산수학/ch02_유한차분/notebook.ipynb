## 이론 정리

### 정의와 관점
- 편미분방정식은 여러 변수 함수의 편도함수 사이 관계로 공간과 시간의 변화를 모델링한다.
- 초기조건은 시간의 시작 상태를, 경계조건은 공간 영역의 가장자리 제약을 정한다.
- 열방정식은 확산과 평활화를, 파동방정식은 전파와 진동을 표현한다.
- 정상상태 문제는 시간 변화가 사라진 공간 분포를 찾는 문제다.

### 핵심 명제와 정리
- 최대원리는 열방정식과 Laplace 방정식에서 내부 값이 경계의 영향을 받는 방식을 설명한다.
- 에너지 방법은 해의 안정성과 유일성을 보이는 핵심 도구다.
- Fourier 방법은 선형 PDE를 주파수 모드별 ODE로 분리한다.
- 유한차분법은 미분연산자를 격자 위 차분연산자로 바꾸어 계산 가능하게 한다.

### 계산과 학습 절차
- 방정식 유형을 elliptic, parabolic, hyperbolic 중 어디에 가까운지 분류한다.
- 영역, 초기조건, 경계조건, 보존량을 먼저 적는다.
- 수치해를 쓸 때는 격자 간격과 시간 간격의 안정조건을 확인한다.
- 해석해가 있을 때 수치해와 비교해 오차와 안정성을 검산한다.

### 자주 생기는 오해
- 경계조건을 바꾸면 같은 방정식도 전혀 다른 해를 갖는다.
- 안정조건을 어긴 유한차분은 그럴듯한 그림을 내더라도 물리적으로 의미가 없다.
- PDE의 해는 고전해, 약해, 분포해처럼 해의 의미가 달라질 수 있다.

### 증명으로 연결하기
- PDE 증명은 적분 부분적분, 에너지 추정, compactness, 약수렴을 자주 사용한다.
- 유일성은 두 해의 차이를 놓고 에너지가 0이 됨을 보이는 방식이 많다.
- 존재성은 Galerkin 근사나 Fourier 전개처럼 유한차원 근사에서 시작해 극한을 취한다.
