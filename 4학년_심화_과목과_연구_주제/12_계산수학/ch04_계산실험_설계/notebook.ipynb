## 이론 정리

### 정의와 관점
- 연구는 정의, 예제, 정리, 증명, 계산, 반례, 응용 사이의 빈틈을 줄이는 과정이다.
- 좋은 연구질문은 충분히 작고, 검증 가능하며, 기존 이론과 연결된다.
- 재현 가능한 연구는 데이터, 코드, 환경, 파라미터, 산출물을 함께 보존한다.
- 논문 읽기는 결과를 암기하는 일이 아니라 가정과 증명 전략을 해체하는 일이다.

### 핵심 명제와 정리
- 문헌조사는 이미 알려진 정리와 아직 열려 있는 질문의 경계를 찾게 해준다.
- 반례는 정리의 가정을 날카롭게 만들고 새로운 조건을 제안하게 한다.
- 계산실험은 추측을 만들고, 증명은 그 추측의 적용 범위를 확정한다.
- 좋은 발표는 문제, 핵심 아이디어, 증거, 한계, 다음 질문을 명확히 연결한다.

### 계산과 학습 절차
- 한 문장 연구질문을 만들고, 필요한 정의와 최소 예제를 정리한다.
- 관련 논문을 표로 정리해 가정, 방법, 결과, 한계를 비교한다.
- 증명 가능한 부분과 실험으로만 확인한 부분을 분리한다.
- 최종 산출물은 노트북, 보고서, 발표자료, 참고문헌을 함께 관리한다.

### 자주 생기는 오해
- 주제가 너무 크면 계산도 증명도 흐려지므로 작은 toy problem으로 시작한다.
- 실험 결과가 흥미로워도 seed와 환경이 없으면 신뢰하기 어렵다.
- 관련 연구와의 차이를 설명하지 못하면 기여가 불분명해진다.

### 증명으로 연결하기
- 졸업연구의 증명은 완전히 새로운 정리보다 기존 정리를 정확히 적용하고 조건을 검산하는 데서 출발해도 좋다.
- 계산 기반 주제는 검증 가능한 benchmark와 오차 분석을 함께 제시해야 한다.
- 글쓰기는 정의, 명제, 증명, 예제, 그림이 서로 같은 이야기를 하도록 정렬하는 작업이다.

- 응용 예제 "논문 결과를 재현 가능한 코드 실험으로 옮긴다."에서 어떤 가정이 현실을 단순화하고 있는가?
- 코드가 연속 대상을 이산화한다면, 격자나 표본 수를 바꾸어도 결론이 유지되는가?

## 0. 실행 준비

아래 셀은 프로젝트 루트의 `common/math_viz.py`를 찾아서 현재 챕터의 출력 폴더를 자동으로 설정합니다. Jupyter Lab을 프로젝트 루트에서 열면 가장 안정적으로 동작합니다.

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("4학년_심화_과목과_연구_주제/12_계산수학/ch04_계산실험_설계")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "capstone_project"
TITLE = "계산수학 - 계산실험 설계"
CONCEPT_EXAMPLE = "입력, 알고리즘, 출력, 평가 기준을 실험 카드로 정리한다."
APPLICATION_EXAMPLE = "논문 결과를 재현 가능한 코드 실험으로 옮긴다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
