## 이론 정리

### 정의와 관점
- 가환대수는 가환환과 아이디얼을 통해 대수방정식의 구조를 연구한다.
- 대수다양체는 다항식들이 동시에 0이 되는 기하적 대상이다.
- 체확대는 방정식의 근을 포함하기 위해 수 체계를 넓히는 과정이다.
- Galois 이론은 다항식 근들의 대칭과 체확대 사이의 대응을 연구한다.

### 핵심 명제와 정리
- Hilbert 기저정리는 다항식환의 아이디얼이 유한 생성됨을 보장한다.
- Nullstellensatz는 대수적으로 닫힌 체에서 아이디얼과 기하적 영점집합을 연결한다.
- Galois 대응은 중간체와 Galois 군의 부분군 사이의 역순 대응을 준다.
- 해의 근호 표현 가능성은 Galois 군의 가해성과 연결된다.

### 계산과 학습 절차
- 다항식 방정식은 아이디얼, 몫환, 영점집합 세 관점으로 번역한다.
- 국소화는 한 점 또는 한 소아이디얼 주변의 정보만 보려는 도구로 이해한다.
- 체확대 문제는 최소다항식, 차수, 켤레근, 분해체를 순서대로 계산한다.
- Galois 군은 근들의 순열 중 대수적 관계를 보존하는 것들로 생각한다.

### 자주 생기는 오해
- 실수 위 기하 직관과 대수적으로 닫힌 체 위 정리는 적용 범위가 다르다.
- 다항식의 근이 추가되면 체의 차수와 대칭 구조가 함께 바뀐다.
- Galois 대응은 정규성, 분리성 같은 조건이 충족될 때 가장 깔끔하게 작동한다.

### 증명으로 연결하기
- 가환대수 증명은 아이디얼 포함관계, 몫, 국소화, 사슬조건을 추적한다.
- Galois 이론 증명은 automorphism이 어떤 원소를 고정하는지 분석한다.
- Groebner 기저는 추상 아이디얼 문제를 계산 가능한 소거 알고리즘으로 바꾼다.
