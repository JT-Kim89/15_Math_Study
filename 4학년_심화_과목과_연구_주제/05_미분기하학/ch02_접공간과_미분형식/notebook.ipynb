## 이론 정리

### 정의와 관점
- 기하학은 공간의 점, 곡선, 곡면, 거리, 각도, 곡률을 연구한다.
- 매개화는 곡선을 시간 또는 좌표 변수로 표현하는 방식이다.
- 곡률은 곡선이나 곡면이 얼마나 빠르게 방향을 바꾸는지 측정한다.
- 리만 계량은 각 점에서 길이와 각도를 재는 규칙을 정한다.

### 핵심 명제와 정리
- 호의 길이와 곡률 공식은 매개화된 곡선의 기하 정보를 계산하게 한다.
- 측지선은 곡면 위에서 국소적으로 가장 곧은 경로이며, 평면의 직선을 일반화한다.
- Gauss의 놀라운 정리는 곡면의 내재적 거리 정보만으로 곡률을 알 수 있음을 말한다.
- 곡률은 삼각형 각의 합, 측지선의 벌어짐, 공간의 전역 형태와 연결된다.

### 계산과 학습 절차
- 매개화, 접벡터, 속도, 법선, 곡률을 순서대로 계산한다.
- 곡면 문제는 좌표계가 특이점을 갖는지 먼저 확인한다.
- 거리 문제는 평면에 그린 그림과 실제 곡면 위 거리를 구분한다.
- 수치 시각화에서는 매개변수 격자의 왜곡과 실제 기하량을 혼동하지 않는다.

### 자주 생기는 오해
- 매개화가 달라져도 곡선 자체는 같을 수 있으므로 좌표 의존량과 기하 불변량을 구분해야 한다.
- 지도 위 직선은 구면 위 최단경로가 아닐 수 있다.
- 곡률 직관은 3차원에 임베딩된 그림에 속기 쉬우므로 내재적 정의를 함께 봐야 한다.

### 증명으로 연결하기
- 기하 증명은 좌표계에서 계산한 뒤 결과가 좌표 선택에 의존하지 않음을 보이는 방식이 많다.
- 측지선 방정식은 에너지 또는 길이 함수의 변분에서 나온다.
- 곡률 정리는 국소 미분량이 전역 형태에 어떤 제약을 주는지를 추적한다.

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("4학년_심화_과목과_연구_주제/05_미분기하학/ch02_접공간과_미분형식")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "geometry_curve"
TITLE = "미분기하학 - 접공간과 미분형식"
CONCEPT_EXAMPLE = "곡선 위 접벡터가 공간에 붙어 움직이는 모습을 그린다."
APPLICATION_EXAMPLE = "로봇이나 드론의 순간 속도 공간을 모델링한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
