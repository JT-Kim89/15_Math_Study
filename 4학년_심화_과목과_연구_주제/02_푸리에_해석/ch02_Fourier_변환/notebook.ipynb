## 이론 정리

### 정의와 관점
- Fourier 해석은 함수를 진동하는 기저들의 합으로 분해하는 관점이다.
- 직교성은 서로 다른 성분이 에너지 관점에서 간섭하지 않음을 뜻한다.
- 함수해석은 벡터공간 개념을 함수공간으로 확장해 노름, 내적, 작용소를 다룬다.
- Hilbert 공간은 완비 내적공간으로, 무한차원에서도 정사영과 직교분해를 가능하게 한다.

### 핵심 명제와 정리
- Parseval 정리는 시간/공간 영역의 에너지와 Fourier 계수의 에너지를 연결한다.
- 합성곱 정리는 시간 영역의 필터링이 주파수 영역의 곱셈과 대응됨을 말한다.
- Riesz 표현정리는 Hilbert 공간의 연속 선형함수를 내적으로 표현한다.
- 스펙트럼 관점은 선형작용소를 주파수 또는 고유모드별로 분해한다.

### 계산과 학습 절차
- 함수를 어떤 기저에 대해 전개할지 정하고 계수 계산식을 쓴다.
- 부분합을 늘리며 근사 오차와 Gibbs 현상 같은 경계 효과를 확인한다.
- 작용소 문제는 정의역, 유계성, 자기수반성, compactness를 점검한다.
- 신호 문제는 시간 영역 그림과 주파수 영역 스펙트럼을 같이 본다.

### 자주 생기는 오해
- Fourier 급수는 불연속점에서 함수값이 아니라 좌우극한 평균으로 수렴할 수 있다.
- 무한차원에서는 모든 선형작용소가 유계인 것은 아니다.
- 점별수렴, L2 수렴, 균등수렴은 서로 다른 개념이다.

### 증명으로 연결하기
- Fourier 정리는 직교기저, 완비성, 수렴 모드의 구분 위에 세워진다.
- 함수해석 증명은 완비성, Banach-Steinhaus, Hahn-Banach 같은 추상 도구를 사용한다.
- PDE 응용에서는 모드별 해를 구한 뒤 수렴성과 경계조건 만족을 확인한다.
