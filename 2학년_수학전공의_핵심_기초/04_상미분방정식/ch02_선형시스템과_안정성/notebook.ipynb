## 이론 정리

### 정의와 관점
- 상미분방정식은 한 변수에 대한 함수와 그 도함수 사이의 관계를 통해 시간 변화를 모델링한다.
- 해는 방정식과 초기조건을 동시에 만족하는 함수다.
- 위상평면은 2차원 시스템의 모든 가능한 궤적을 한 그림에 담는다.
- 안정성은 작은 초기 오차가 시간이 지나며 줄어드는지 커지는지를 묻는다.

### 핵심 명제와 정리
- 존재와 유일성 정리는 적절한 매끄러움 조건에서 초기값 문제가 잘 정의됨을 보장한다.
- 선형시스템의 장기 거동은 행렬의 고유값 실수부와 직접 연결된다.
- 평형점 주변 선형화는 비선형 시스템의 국소 동역학을 이해하게 해준다.
- Euler와 Runge-Kutta 방법은 연속 시간을 작은 단계로 이산화해 해를 근사한다.

### 계산과 학습 절차
- 먼저 평형점, 보존량, 단조성, 양의 불변영역을 찾는다.
- 해석해가 어렵다면 방향장이나 위상평면으로 질적 거동을 본다.
- 수치해는 시간 간격을 줄이며 안정성과 수렴성을 비교한다.
- 모형에서는 변수의 단위, 파라미터의 의미, 초기조건의 현실성을 확인한다.

### 자주 생기는 오해
- 수치해가 매끄러워 보여도 시간 간격이 너무 크면 완전히 잘못된 안정성을 보일 수 있다.
- 선형화는 평형점 근처에서만 믿을 수 있고 전역 거동을 보장하지 않는다.
- 모든 ODE가 닫힌형 해를 갖는 것은 아니므로 질적 분석과 수치분석이 중요하다.

### 증명으로 연결하기
- 존재정리는 보통 적분방정식으로 바꾸고 고정점 정리를 적용하는 방식으로 증명된다.
- 안정성 증명은 Lyapunov 함수처럼 에너지 역할을 하는 양을 찾는 것이 핵심이다.
- 수치해법 증명은 국소절단오차와 오차 누적을 분리해 본다.

## 0. 실행 준비

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("2학년_수학전공의_핵심_기초/04_상미분방정식/ch02_선형시스템과_안정성")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "ode_phase"
TITLE = "상미분방정식 - 선형시스템과 안정성"
CONCEPT_EXAMPLE = "고정점 주변 궤적이 안정/불안정하게 움직이는지 본다."
APPLICATION_EXAMPLE = "간단한 경제 균형 모형이 충격 후 회복되는지 분석한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
