## 이론 정리

### 정의와 관점
- 응용수학은 현실 시스템을 변수, 방정식, 확률모형, 최적화 문제로 번역한다.
- 수리물리는 대칭, 보존량, 변분원리로 물리 현상을 설명한다.
- 금융수학은 시간가치, 무차익, 확률과정을 이용해 현금흐름과 위험을 모델링한다.
- 모형은 현실의 전체가 아니라 질문에 필요한 구조만 남긴 의도적 단순화다.

### 핵심 명제와 정리
- 변분원리는 많은 물리 법칙을 어떤 함수의 최적화 조건으로 표현한다.
- 보존량은 시스템의 장기 거동과 수치해 검증에 강한 기준을 준다.
- 무차익 원리는 금융상품 가격이 서로 모순되지 않도록 하는 핵심 조건이다.
- 위험측도는 평균만으로 보이지 않는 꼬리 손실을 정량화한다.

### 계산과 학습 절차
- 현실 문제에서 변수, 단위, 가정, 보존량, 관측 가능한 출력을 정리한다.
- 모형을 세운 뒤 특수한 경우와 극한 경우가 상식에 맞는지 확인한다.
- 시뮬레이션은 파라미터 민감도와 불확실성 범위를 함께 보여준다.
- 금융 모형은 기대수익보다 위험, 할인, 헤지 가능성을 함께 본다.

### 자주 생기는 오해
- 모형이 맞아 보이는 것은 데이터 범위 안에서만일 수 있으며 외삽은 위험하다.
- 물리적 단위를 무시하면 수식이 차원적으로 틀릴 수 있다.
- 금융의 정규분포 가정은 극단 손실을 과소평가할 수 있다.

### 증명으로 연결하기
- 응용수학의 정당화는 가정에서 방정식을 유도하고, 방정식에서 예측을 뽑아 데이터와 비교하는 순서다.
- 변분 문제는 작은 경로 변화에 대한 1차 변화량을 0으로 두어 Euler-Lagrange 방정식을 얻는다.
- 금융수학 증명은 복제 포트폴리오, martingale measure, 할인 기대값을 중심으로 전개된다.

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("2학년_수학전공의_핵심_기초/04_상미분방정식/ch03_진동과_강제응답")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "math_physics"
TITLE = "상미분방정식 - 진동과 강제응답"
CONCEPT_EXAMPLE = "질량-스프링 시스템의 위치와 에너지를 시각화한다."
APPLICATION_EXAMPLE = "건물 진동의 공진 위험을 간단한 모형으로 이해한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
