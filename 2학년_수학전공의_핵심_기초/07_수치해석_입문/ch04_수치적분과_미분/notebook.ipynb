## 이론 정리

### 정의와 관점
- 극한은 함수값 자체보다 입력이 한 점에 가까워질 때의 안정적인 경향을 다룬다.
- 연속성은 작은 입력 변화가 작은 출력 변화로 이어진다는 국소 안정성의 수학적 표현이다.
- 미분은 국소적인 선형 근사이고, 적분은 작은 조각들의 누적을 극한으로 보낸 것이다.
- 급수와 Taylor 전개는 복잡한 함수를 다항식이나 단순한 항들의 합으로 바꾸는 근사 언어다.

### 핵심 명제와 정리
- 평균값정리는 미분가능한 함수의 전역 변화량과 어떤 한 점의 순간 변화율을 연결한다.
- 미적분학의 기본정리는 누적량의 변화율과 원래 함수가 서로 역방향으로 연결됨을 말한다.
- Taylor 정리는 함수값, 도함수값, 나머지항을 함께 사용해 근사의 정확도를 통제한다.
- 수렴판정법은 무한합을 실제 계산 가능한 유한 정보로 판단하게 해준다.

### 계산과 학습 절차
- 정의역과 특이점을 먼저 찾고, 극한/연속/미분/적분 중 어떤 관점이 필요한지 고른다.
- 작은 예제에서는 손계산으로 극한값, 도함수, 적분값을 먼저 구한 뒤 수치 결과와 비교한다.
- 연속 대상을 코드로 다룰 때는 격자 간격을 줄이며 결과가 안정되는지 확인한다.
- 근사식은 항상 오차항이나 잔차 그래프를 함께 보며 어디서 깨지는지 확인한다.

### 자주 생기는 오해
- 그래프가 매끄러워 보여도 미분가능하지 않을 수 있고, 점 하나의 값은 극한을 결정하지 않는다.
- 무한급수의 항이 0으로 가는 것은 수렴의 필요조건일 뿐 충분조건이 아니다.
- 수치 적분과 미분은 격자 선택, 반올림, 함수의 진동성에 민감하다.

### 증명으로 연결하기
- epsilon-delta 정의는 직관을 문장으로 고정하는 도구이므로 먼저 그림으로 근방 관계를 해석한다.
- 존재정리는 보통 완비성, compactness, 연속성 같은 조건을 이용해 극한점을 확보한다.
- 계산 공식은 정의에서 출발해 선형성, 극한 교환, 오차 추정을 통해 정당화한다.

```python
from pathlib import Path
import sys
from IPython.display import Image, display


CHAPTER_RELATIVE_DIR = Path("2학년_수학전공의_핵심_기초/07_수치해석_입문/ch04_수치적분과_미분")


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "common" / "math_viz.py").exists():
            return candidate
    raise RuntimeError("common/math_viz.py를 찾지 못했습니다. Jupyter Lab을 프로젝트 루트에서 열어 주세요.")


ROOT = find_project_root(Path.cwd().resolve())
NOTEBOOK_DIR = ROOT / CHAPTER_RELATIVE_DIR
OUTPUT_DIR = NOTEBOOK_DIR / "outputs"
sys.path.insert(0, str(ROOT / "common"))

from math_viz import PROFILES, run_profile

PROFILE = "calculus_integral"
TITLE = "수치해석 입문 - 수치적분과 미분"
CONCEPT_EXAMPLE = "사다리꼴 합과 직사각형 합을 비교한다."
APPLICATION_EXAMPLE = "속도 데이터로 이동거리를 계산한다."

print("project root:", ROOT)
print("chapter dir:", NOTEBOOK_DIR)
print("profile:", PROFILE)
```
